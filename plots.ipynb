```python
import numpy as np
import matplotlib.pyplot as plt
import json
```

```python
with open('training_history.json', 'r') as f:
    data = json.load(f)
```

```python
epochs = data['epochs']
losses = data['losses']
val_mrr = data['val_mrr']
val_hits10 = data['val_hits10']
```

```python
plt.figure(figsize=(12, 8))
plt.plot(epochs, losses, color='orange', marker='^')
plt.xlabel("Epochs")
plt.ylabel("Validation Loss")
plt.title("Model Training Loss Curve")
plt.grid()
```

```python
plt.figure(figsize=(12, 8))
plt.plot(epochs, val_mrr, color='green', marker='*')
plt.xlabel("Epochs")
plt.ylabel("Validation MRR")
plt.title("Model Training Validation MRR Curve")
plt.grid()
```

```python
plt.figure(figsize=(12, 8))
plt.plot(epochs, val_hits10, color='blue', marker='.')
plt.xlabel("Epochs")
plt.ylabel("Hits@10")
plt.title("Model Training Validation Hits@10 Curve")
plt.grid()
```
